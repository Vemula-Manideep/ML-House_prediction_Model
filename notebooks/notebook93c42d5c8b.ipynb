```python
class StandardScalerFromScratch:

    def fit(self, X):
        self.mean = np.mean(X, axis=0)
        self.std = np.std(X, axis=0)

        # Prevent division by zero
        self.std[self.std == 0] = 1

        return self

    def transform(self, X):
        return (X - self.mean) / self.std

    def fit_transform(self, X):
        self.fit(X)
        return self.transform(X)
```

```python
def compute_gradient(X, y, weights):
    m = X.shape[0]

    predictions = X @ weights

    gradient = (2 / m) * X.T @ (predictions - y)

    return gradient
```

```python
class LinearRegressionFromScratch:

    def __init__(
        self,
        learning_rate=0.001,
        n_iterations=1000
    ):
        self.learning_rate = learning_rate
        self.n_iterations = n_iterations
        self.weights = None
        self.loss_history = []

    def fit(self, X, y):

        # Number of features
        n_features = X.shape[1]

        # Initialize weights
        self.weights = np.zeros(n_features)

        # Gradient Descent
        for i in range(self.n_iterations):

            # Predictions
            predictions = X @ self.weights

            # Error
            error = predictions - y

            # MSE
            loss = np.mean(error ** 2)

            # Store loss
            self.loss_history.append(loss)

            # Gradient
            gradient = (
                (2 / X.shape[0])
                * X.T
                @ error
            )

            # Update weights
            self.weights -= (
                self.learning_rate * gradient
            )

        return self

    def predict(self, X):

        return X @ self.weights
```

```python
model = LinearRegressionFromScratch(
    learning_rate=0.001,
    n_iterations=2000
)

model.fit(
    X_train_bias,
    y_train
)
```

```python
import numpy as np
import pandas as pd

# -----------------------------
# Metric functions from scratch
# -----------------------------

def mean_absolute_error(y_true, y_pred):
    return np.mean(np.abs(y_true - y_pred))


def mean_squared_error(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)


def root_mean_squared_error(y_true, y_pred):
    return np.sqrt(
        mean_squared_error(y_true, y_pred)
    )


def r2_score(y_true, y_pred):
    ss_res = np.sum((y_true - y_pred) ** 2)
    ss_total = np.sum(
        (y_true - np.mean(y_true)) ** 2
    )

    return 1 - (ss_res / ss_total)


# -----------------------------
# TRAIN metrics
# -----------------------------

train_mae = mean_absolute_error(
    y_train,
    train_predictions
)

train_mse = mean_squared_error(
    y_train,
    train_predictions
)

train_rmse = root_mean_squared_error(
    y_train,
    train_predictions
)

train_r2 = r2_score(
    y_train,
    train_predictions
)


# -----------------------------
# TEST metrics
# -----------------------------

test_mae = mean_absolute_error(
    y_test,
    test_predictions
)

test_mse = mean_squared_error(
    y_test,
    test_predictions
)

test_rmse = root_mean_squared_error(
    y_test,
    test_predictions
)

test_r2 = r2_score(
    y_test,
    test_predictions
)


# -----------------------------
# Display results
# -----------------------------

metrics = pd.DataFrame({
    "Metric": [
        "MAE",
        "MSE",
        "RMSE",
        "R²"
    ],
    "Train": [
        train_mae,
        train_mse,
        train_rmse,
        train_r2
    ],
    "Test": [
        test_mae,
        test_mse,
        test_rmse,
        test_r2
    ]
})

print(metrics)
```

```python
class RidgeRegressionFromScratch:

    def __init__(
        self,
        learning_rate=0.001,
        n_iterations=2000,
        lambda_=0.1
    ):
        self.learning_rate = learning_rate
        self.n_iterations = n_iterations
        self.lambda_ = lambda_

        self.weights = None
        self.loss_history = []

    def fit(self, X, y):

        n_features = X.shape[1]

        # Initialize weights
        self.weights = np.zeros(n_features)

        for i in range(self.n_iterations):

            # -------------------------
            # 1. Predictions
            # -------------------------

            predictions = X @ self.weights

            # -------------------------
            # 2. Error
            # -------------------------

            error = predictions - y

            # -------------------------
            # 3. MSE
            # -------------------------

            mse = np.mean(error ** 2)

            # -------------------------
            # 4. Ridge penalty
            # Don't penalize bias
            # -------------------------

            penalty = (
                self.lambda_
                * np.sum(self.weights[1:] ** 2)
            )

            # -------------------------
            # 5. Total loss
            # -------------------------

            loss = mse + penalty

            self.loss_history.append(loss)

            # -------------------------
            # 6. Gradient
            # -------------------------

            gradient = (
                (2 / X.shape[0])
                * X.T
                @ error
            )

            # -------------------------
            # 7. Ridge regularization
            # Don't regularize bias
            # -------------------------

            gradient[1:] += (
                2
                * self.lambda_
                * self.weights[1:]
            )

            # -------------------------
            # 8. Update weights
            # -------------------------

            self.weights -= (
                self.learning_rate
                * gradient
            )

        return self

    def predict(self, X):

        return X @ self.weights
```

```python
class LassoRegressionFromScratch:

    def __init__(
        self,
        learning_rate=0.001,
        n_iterations=2000,
        lambda_=0.001
    ):
        self.learning_rate = learning_rate
        self.n_iterations = n_iterations
        self.lambda_ = lambda_

        self.weights = None
        self.loss_history = []

    def fit(self, X, y):

        n_features = X.shape[1]

        # Initialize weights
        self.weights = np.zeros(n_features)

        for i in range(self.n_iterations):

            # -------------------------
            # 1. Predictions
            # -------------------------

            predictions = X @ self.weights

            # -------------------------
            # 2. Error
            # -------------------------

            error = predictions - y

            # -------------------------
            # 3. MSE
            # -------------------------

            mse = np.mean(error ** 2)

            # -------------------------
            # 4. L1 penalty
            # Don't penalize bias
            # -------------------------

            penalty = (
                self.lambda_
                * np.sum(np.abs(self.weights[1:]))
            )

            # -------------------------
            # 5. Total loss
            # -------------------------

            loss = mse + penalty

            self.loss_history.append(loss)

            # -------------------------
            # 6. MSE gradient
            # -------------------------

            gradient = (
                (2 / X.shape[0])
                * X.T
                @ error
            )

            # -------------------------
            # 7. L1 gradient
            # -------------------------

            gradient[1:] += (
                self.lambda_
                * np.sign(self.weights[1:])
            )

            # -------------------------
            # 8. Update weights
            # -------------------------

            self.weights -= (
                self.learning_rate
                * gradient
            )

        return self

    def predict(self, X):

        return X @ self.weights
```

```python
def k_fold_split(X, y, k=5, random_state=42):

    np.random.seed(random_state)

    n_samples = X.shape[0]

    # Shuffle indices
    indices = np.arange(n_samples)
    np.random.shuffle(indices)

    # Split indices into k folds
    folds = np.array_split(indices, k)

    for i in range(k):

        validation_indices = folds[i]

        training_indices = np.concatenate(
            [folds[j] for j in range(k) if j != i]
        )

        X_train_fold = X[training_indices]
        X_val_fold = X[validation_indices]

        y_train_fold = y[training_indices]
        y_val_fold = y[validation_indices]

        yield (
            X_train_fold,
            X_val_fold,
            y_train_fold,
            y_val_fold
        )
```

```python
fold_results = []

for fold, (
    X_train_fold,
    X_val_fold,
    y_train_fold,
    y_val_fold
) in enumerate(
    k_fold_split(
        X_train_bias,
        y_train,
        k=5,
        random_state=42
    ),
    start=1
):

    model = LinearRegressionFromScratch(
        learning_rate=0.001,
        n_iterations=2000
    )

    model.fit(
        X_train_fold,
        y_train_fold
    )

    predictions = model.predict(
        X_val_fold
    )

    mae = mean_absolute_error(
        y_val_fold,
        predictions
    )

    rmse = root_mean_squared_error(
        y_val_fold,
        predictions
    )

    r2 = r2_score(
        y_val_fold,
        predictions
    )

    fold_results.append({
        "Fold": fold,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    })

cv_results = pd.DataFrame(fold_results)

print(cv_results)
```

```python
cv_results = []

for fold, (train_idx, val_idx) in enumerate(
    kf.split(X),
    start=1
):

    print(f"\n========== Fold {fold} ==========")

    # -------------------------
    # Split RAW data
    # -------------------------

    X_train_fold = X.iloc[train_idx]
    X_val_fold = X.iloc[val_idx]

    y_train_fold = y.iloc[train_idx]
    y_val_fold = y.iloc[val_idx]

    # -------------------------
    # Fit preprocessing ONLY
    # on training fold
    # -------------------------

    X_train_processed = (
        preprocessor.fit_transform(
            X_train_fold
        )
    )

    # -------------------------
    # Transform validation
    # using training parameters
    # -------------------------

    X_val_processed = (
        preprocessor.transform(
            X_val_fold
        )
    )

    # Convert to NumPy
    X_train_processed = np.asarray(
        X_train_processed,
        dtype=np.float64
    )

    X_val_processed = np.asarray(
        X_val_processed,
        dtype=np.float64
    )

    y_train_fold = np.asarray(
        y_train_fold,
        dtype=np.float64
    )

    y_val_fold = np.asarray(
        y_val_fold,
        dtype=np.float64
    )

    # -------------------------
    # Scale
    # -------------------------

    scaler = StandardScalerFromScratch()

    X_train_scaled = scaler.fit_transform(
        X_train_processed
    )

    X_val_scaled = scaler.transform(
        X_val_processed
    )

    # -------------------------
    # Add bias
    # -------------------------

    X_train_bias_fold = np.c_[
        np.ones(X_train_scaled.shape[0]),
        X_train_scaled
    ]

    X_val_bias_fold = np.c_[
        np.ones(X_val_scaled.shape[0]),
        X_val_scaled
    ]

    # -------------------------
    # Train from scratch
    # -------------------------

    model = LinearRegressionFromScratch(
        learning_rate=0.001,
        n_iterations=2000
    )

    model.fit(
        X_train_bias_fold,
        y_train_fold
    )

    # -------------------------
    # Predict
    # -------------------------

    predictions = model.predict(
        X_val_bias_fold
    )

    # -------------------------
    # Metrics
    # -------------------------

    mae = mean_absolute_error(
        y_val_fold,
        predictions
    )

    rmse = root_mean_squared_error(
        y_val_fold,
        predictions
    )

    r2 = r2_score(
        y_val_fold,
        predictions
    )

    cv_results.append({
        "Fold": fold,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    })

cv_results_df = pd.DataFrame(cv_results)

print("\n")
print(cv_results_df)
```

```python
# ============================================================
# HOUSE PRICE PREDICTION
# COMPLETE ML PIPELINE AFTER FEATURE ENGINEERING
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import KFold
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder


# ============================================================
# 1. LOAD DATA
# ============================================================

train_path = (
    "/kaggle/input/datasets/rishitaverma02/"
    "house-prices-advanced-regression-techniques/train (1).csv"
)

test_path = (
    "/kaggle/input/datasets/rishitaverma02/"
    "house-prices-advanced-regression-techniques/test (1).csv"
)

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

print("Train shape:", train_df.shape)
print("Test shape :", test_df.shape)


# ============================================================
# 2. SEPARATE TARGET
# ============================================================

X = train_df.drop(columns=["SalePrice"]).copy()
y = train_df["SalePrice"].copy()

X_test = test_df.copy()

# Remove Id from model features
X = X.drop(columns=["Id"], errors="ignore")
X_test = X_test.drop(columns=["Id"], errors="ignore")

print("\nX shape:", X.shape)
print("y shape:", y.shape)
print("Test shape:", X_test.shape)


# ============================================================
# 3. FEATURE ENGINEERING FUNCTION
# ============================================================

def create_features(df):
    
    df = df.copy()

    # --------------------------------------------------------
    # Total square footage
    # --------------------------------------------------------

    df["TotalSF"] = (
        df["TotalBsmtSF"].fillna(0)
        + df["1stFlrSF"].fillna(0)
        + df["2ndFlrSF"].fillna(0)
    )

    # --------------------------------------------------------
    # House age
    # --------------------------------------------------------

    df["HouseAge"] = (
        df["YrSold"]
        - df["YearBuilt"]
    )

    # --------------------------------------------------------
    # Remodel age
    # --------------------------------------------------------

    df["RemodAge"] = (
        df["YrSold"]
        - df["YearRemodAdd"]
    )

    # --------------------------------------------------------
    # Total bathrooms
    # --------------------------------------------------------

    df["TotalBathrooms"] = (
        df["FullBath"].fillna(0)
        + 0.5 * df["HalfBath"].fillna(0)
        + df["BsmtFullBath"].fillna(0)
        + 0.5 * df["BsmtHalfBath"].fillna(0)
    )

    # --------------------------------------------------------
    # Total porch/deck area
    # --------------------------------------------------------

    df["TotalPorchSF"] = (
        df["OpenPorchSF"].fillna(0)
        + df["3SsnPorch"].fillna(0)
        + df["EnclosedPorch"].fillna(0)
        + df["ScreenPorch"].fillna(0)
        + df["WoodDeckSF"].fillna(0)
    )

    # --------------------------------------------------------
    # Total finished basement
    # --------------------------------------------------------

    df["TotalBsmtFinSF"] = (
        df["BsmtFinSF1"].fillna(0)
        + df["BsmtFinSF2"].fillna(0)
    )

    # --------------------------------------------------------
    # Above-ground area
    # --------------------------------------------------------

    df["TotalAboveGroundSF"] = (
        df["1stFlrSF"].fillna(0)
        + df["2ndFlrSF"].fillna(0)
    )

    # --------------------------------------------------------
    # Quality × living area
    # --------------------------------------------------------

    df["OverallQual_GrLivArea"] = (
        df["OverallQual"]
        * df["GrLivArea"]
    )

    # --------------------------------------------------------
    # Quality × total area
    # --------------------------------------------------------

    df["OverallQual_TotalSF"] = (
        df["OverallQual"]
        * df["TotalSF"]
    )

    return df


# ============================================================
# 4. APPLY FEATURE ENGINEERING
# ============================================================

X_fe = create_features(X)
X_test_fe = create_features(X_test)

print("\nAfter feature engineering:")
print("Training:", X_fe.shape)
print("Testing :", X_test_fe.shape)


# ============================================================
# 5. SHOW NEW FEATURES
# ============================================================

new_features = [
    "TotalSF",
    "HouseAge",
    "RemodAge",
    "TotalBathrooms",
    "TotalPorchSF",
    "TotalBsmtFinSF",
    "TotalAboveGroundSF",
    "OverallQual_GrLivArea",
    "OverallQual_TotalSF"
]

print("\nNew features:")
print(X_fe[new_features].head())


# ============================================================
# 6. IDENTIFY COLUMN TYPES
# ============================================================

categorical_columns = X_fe.select_dtypes(
    include=["object"]
).columns.tolist()

numerical_columns = X_fe.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

print("\nNumerical columns:", len(numerical_columns))
print("Categorical columns:", len(categorical_columns))


# ============================================================
# 7. CREATE PREPROCESSOR
# ============================================================

def create_preprocessor(
    numerical_columns,
    categorical_columns
):

    numerical_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ])

    categorical_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ])

    preprocessor = ColumnTransformer([
        (
            "numerical",
            numerical_pipeline,
            numerical_columns
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_columns
        )
    ])

    return preprocessor


# ============================================================
# 8. LEAKAGE-FREE CROSS VALIDATION
# ============================================================

kf = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


def evaluate_linear_model_cv(
    X_data,
    y_data,
    n_splits=5,
    random_state=42
):

    kf = KFold(
        n_splits=n_splits,
        shuffle=True,
        random_state=random_state
    )

    results = []

    for fold, (train_idx, val_idx) in enumerate(
        kf.split(X_data),
        start=1
    ):

        print(f"\n========== Fold {fold} ==========")

        X_train_fold = X_data.iloc[train_idx]
        X_val_fold = X_data.iloc[val_idx]

        y_train_fold = y_data.iloc[train_idx]
        y_val_fold = y_data.iloc[val_idx]

        # ----------------------------------------------------
        # Identify columns from training fold
        # ----------------------------------------------------

        numerical_columns_fold = (
            X_train_fold
            .select_dtypes(
                include=["int64", "float64"]
            )
            .columns
            .tolist()
        )

        categorical_columns_fold = (
            X_train_fold
            .select_dtypes(
                include=["object"]
            )
            .columns
            .tolist()
        )

        # ----------------------------------------------------
        # Create NEW preprocessor for each fold
        # ----------------------------------------------------

        preprocessor = create_preprocessor(
            numerical_columns_fold,
            categorical_columns_fold
        )

        # ----------------------------------------------------
        # Fit only on training fold
        # ----------------------------------------------------

        X_train_processed = (
            preprocessor.fit_transform(
                X_train_fold
            )
        )

        # ----------------------------------------------------
        # Transform validation fold
        # ----------------------------------------------------

        X_val_processed = (
            preprocessor.transform(
                X_val_fold
            )
        )

        # ----------------------------------------------------
        # NumPy conversion
        # ----------------------------------------------------

        X_train_processed = np.asarray(
            X_train_processed,
            dtype=np.float64
        )

        X_val_processed = np.asarray(
            X_val_processed,
            dtype=np.float64
        )

        y_train_array = np.asarray(
            y_train_fold,
            dtype=np.float64
        )

        y_val_array = np.asarray(
            y_val_fold,
            dtype=np.float64
        )

        # ----------------------------------------------------
        # Scaling
        # ----------------------------------------------------

        scaler = StandardScalerFromScratch()

        X_train_scaled = scaler.fit_transform(
            X_train_processed
        )

        X_val_scaled = scaler.transform(
            X_val_processed
        )

        # ----------------------------------------------------
        # Add bias
        # ----------------------------------------------------

        X_train_bias = np.c_[
            np.ones(X_train_scaled.shape[0]),
            X_train_scaled
        ]

        X_val_bias = np.c_[
            np.ones(X_val_scaled.shape[0]),
            X_val_scaled
        ]

        # ----------------------------------------------------
        # Train
        # ----------------------------------------------------

        model = LinearRegressionFromScratch(
            learning_rate=0.001,
            n_iterations=2000
        )

        model.fit(
            X_train_bias,
            y_train_array
        )

        # ----------------------------------------------------
        # Predict
        # ----------------------------------------------------

        predictions = model.predict(
            X_val_bias
        )

        # ----------------------------------------------------
        # Metrics
        # ----------------------------------------------------

        mae = mean_absolute_error(
            y_val_array,
            predictions
        )

        mse = mean_squared_error(
            y_val_array,
            predictions
        )

        rmse = root_mean_squared_error(
            y_val_array,
            predictions
        )

        r2 = r2_score(
            y_val_array,
            predictions
        )

        results.append({
            "Fold": fold,
            "MAE": mae,
            "MSE": mse,
            "RMSE": rmse,
            "R2": r2
        })

    return pd.DataFrame(results)


# ============================================================
# 9. RUN FEATURE-ENGINEERED CV
# ============================================================

fe_cv_results = evaluate_linear_model_cv(
    X_fe,
    y,
    n_splits=5,
    random_state=42
)

print("\n")
print("=" * 60)
print("FEATURE ENGINEERED CROSS-VALIDATION")
print("=" * 60)

print(fe_cv_results)


# ============================================================
# 10. FEATURE ENGINEERING SUMMARY
# ============================================================

fe_mean_mae = fe_cv_results["MAE"].mean()
fe_mean_mse = fe_cv_results["MSE"].mean()
fe_mean_rmse = fe_cv_results["RMSE"].mean()
fe_std_rmse = fe_cv_results["RMSE"].std()
fe_mean_r2 = fe_cv_results["R2"].mean()
fe_std_r2 = fe_cv_results["R2"].std()

print("\n===== FEATURE ENGINEERING SUMMARY =====")

print(f"Mean MAE  : ${fe_mean_mae:,.2f}")
print(f"Mean MSE  : {fe_mean_mse:,.2f}")
print(f"Mean RMSE : ${fe_mean_rmse:,.2f}")
print(f"RMSE Std  : ${fe_std_rmse:,.2f}")
print(f"Mean R²   : {fe_mean_r2:.4f}")
print(f"R² Std    : {fe_std_r2:.4f}")


# ============================================================
# 11. COMPARE BASELINE VS FEATURE ENGINEERING
# ============================================================

# Your previously obtained leakage-free baseline
baseline_mean_mae = 18354.35
baseline_mean_rmse = 34247.20
baseline_mean_r2 = 0.7923

comparison = pd.DataFrame({

    "Model": [
        "Baseline Linear Regression",
        "Feature Engineered Linear Regression"
    ],

    "Mean MAE": [
        baseline_mean_mae,
        fe_mean_mae
    ],

    "Mean RMSE": [
        baseline_mean_rmse,
        fe_mean_rmse
    ],

    "Mean R2": [
        baseline_mean_r2,
        fe_mean_r2
    ]
})

print("\n")
print("=" * 60)
print("BASELINE VS FEATURE ENGINEERING")
print("=" * 60)

print(comparison)


# ============================================================
# 12. CALCULATE IMPROVEMENT
# ============================================================

rmse_improvement = (
    (baseline_mean_rmse - fe_mean_rmse)
    / baseline_mean_rmse
) * 100

r2_improvement = (
    fe_mean_r2 - baseline_mean_r2
)

print("\n===== IMPROVEMENT =====")

print(
    f"RMSE improvement: "
    f"{rmse_improvement:.2f}%"
)

print(
    f"R² improvement: "
    f"{r2_improvement:.4f}"
)


# ============================================================
# 13. SELECT BEST FEATURE SET
# ============================================================

if fe_mean_rmse < baseline_mean_rmse:

    print(
        "\nFeature engineering IMPROVED "
        "the model."
    )

    use_feature_engineering = True

else:

    print(
        "\nFeature engineering DID NOT "
        "improve the model."
    )

    use_feature_engineering = False


# ============================================================
# 14. FINAL DATASET
# ============================================================

if use_feature_engineering:

    X_final = X_fe.copy()
    X_test_final = X_test_fe.copy()

else:

    X_final = X.copy()
    X_test_final = X_test.copy()


print("\nFinal training shape:", X_final.shape)
print("Final test shape:", X_test_final.shape)


# ============================================================
# 15. FINAL PREPROCESSING
# ============================================================

numerical_columns_final = X_final.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_columns_final = X_final.select_dtypes(
    include=["object"]
).columns.tolist()

final_preprocessor = create_preprocessor(
    numerical_columns_final,
    categorical_columns_final
)


# ============================================================
# 16. FIT PREPROCESSOR ON FULL TRAINING DATA
# ============================================================

X_final_processed = (
    final_preprocessor.fit_transform(
        X_final
    )
)

X_test_final_processed = (
    final_preprocessor.transform(
        X_test_final
    )
)

X_final_processed = np.asarray(
    X_final_processed,
    dtype=np.float64
)

X_test_final_processed = np.asarray(
    X_test_final_processed,
    dtype=np.float64
)

y_array = np.asarray(
    y,
    dtype=np.float64
)


# ============================================================
# 17. FINAL SCALING
# ============================================================

final_scaler = StandardScalerFromScratch()

X_final_scaled = final_scaler.fit_transform(
    X_final_processed
)

X_test_final_scaled = final_scaler.transform(
    X_test_final_processed
)


# ============================================================
# 18. ADD BIAS
# ============================================================

X_final_bias = np.c_[
    np.ones(X_final_scaled.shape[0]),
    X_final_scaled
]

X_test_final_bias = np.c_[
    np.ones(X_test_final_scaled.shape[0]),
    X_test_final_scaled
]


print("\nFinal matrix:")
print("Train:", X_final_bias.shape)
print("Test :", X_test_final_bias.shape)


# ============================================================
# 19. TRAIN FINAL LINEAR REGRESSION
# ============================================================

final_model = LinearRegressionFromScratch(
    learning_rate=0.001,
    n_iterations=2000
)

final_model.fit(
    X_final_bias,
    y_array
)


# ============================================================
# 20. TRAINING PREDICTIONS
# ============================================================

final_train_predictions = (
    final_model.predict(
        X_final_bias
    )
)

final_test_predictions = (
    final_model.predict(
        X_test_final_bias
    )
)


# ============================================================
# 21. TRAINING METRICS
# ============================================================

final_train_mae = mean_absolute_error(
    y_array,
    final_train_predictions
)

final_train_mse = mean_squared_error(
    y_array,
    final_train_predictions
)

final_train_rmse = root_mean_squared_error(
    y_array,
    final_train_predictions
)

final_train_r2 = r2_score(
    y_array,
    final_train_predictions
)


print("\n")
print("=" * 60)
print("FINAL MODEL TRAINING PERFORMANCE")
print("=" * 60)

print(f"MAE  : ${final_train_mae:,.2f}")
print(f"MSE  : {final_train_mse:,.2f}")
print(f"RMSE : ${final_train_rmse:,.2f}")
print(f"R²   : {final_train_r2:.4f}")


# ============================================================
# 22. PREDICTION SANITY CHECK
# ============================================================

print("\nFirst 10 predictions:")

prediction_check = pd.DataFrame({
    "Prediction": final_test_predictions[:10]
})

print(prediction_check)


# ============================================================
# 23. CHECK INVALID PREDICTIONS
# ============================================================

print("\n===== PREDICTION CHECK =====")

print(
    "NaN predictions:",
    np.isnan(final_test_predictions).sum()
)

print(
    "Infinite predictions:",
    np.isinf(final_test_predictions).sum()
)

print(
    "Minimum prediction:",
    final_test_predictions.min()
)

print(
    "Maximum prediction:",
    final_test_predictions.max()
)


# ============================================================
# 24. OPTIONAL: CLIP NEGATIVE HOUSE PRICES
# ============================================================

final_test_predictions = np.maximum(
    final_test_predictions,
    0
)


# ============================================================
# 25. CREATE KAGGLE SUBMISSION
# ============================================================

submission = pd.DataFrame({

    "Id": test_df["Id"],

    "SalePrice": final_test_predictions

})


print("\nSubmission preview:")
print(submission.head())


# ============================================================
# 26. CHECK SUBMISSION
# ============================================================

print("\nSubmission shape:")
print(submission.shape)

print("\nMissing values:")
print(submission.isnull().sum())


# ============================================================
# 27. SAVE SUBMISSION
# ============================================================

submission_path = (
    "/kaggle/working/"
    "submission_linear_regression.csv"
)

submission.to_csv(
    submission_path,
    index=False
)

print(
    "\nSubmission saved to:"
)

print(submission_path)


# ============================================================
# 28. VERIFY FILE
# ============================================================

print("\nSaved file exists:")

print(
    os.path.exists(
        submission_path
    )
)


# ============================================================
# 29. FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 60)
print("PROJECT SUMMARY")
print("=" * 60)

print("Model: Linear Regression from Scratch")

if use_feature_engineering:
    print("Feature Engineering: ENABLED")
else:
    print("Feature Engineering: NOT USED")

print(
    f"CV Mean RMSE: "
    f"${fe_mean_rmse:,.2f}"
)

print(
    f"CV Mean R²: "
    f"{fe_mean_r2:.4f}"
)

print(
    f"CV RMSE Std: "
    f"${fe_std_rmse:,.2f}"
)

print(
    "\nSubmission:"
)

print(submission_path)
```
